```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from tensorflow.keras import layers
import pandas as pd
```

```python
df = pd.DataFrame({ 'soil_moisture': [12, 18, 22, 28, 35, 40, 15, 20, 25, 30, 14, 32, 38, 17, 27, 42],
                   'temperature_C': [36, 34, 31, 29, 27, 25, 37, 33, 30, 28, 35, 26, 24, 32, 29, 23],
                    'sunlight_hours': [11, 10, 8, 7, 6, 5, 12, 9, 8, 7, 11, 6, 5, 9, 7, 4],
                    'needs_water': [1, 1, 1, 0, 0, 0, 1, 1, 0, 0, 1, 0, 0, 1, 0, 0] })
```

```python
df
```

```python
df.columns
```

```python
X = df[['soil_moisture', 'temperature_C', 'sunlight_hours']]
y = df['needs_water']
# we need to scale these values  for temperature for values between 0 and 1 which is normalization and standardization mean - 0 and sd = 1 with values between -3 and 3
```

```python
# we do normalization here
X_min = X.min()
X_max = X.max()
X_scaled = (X-X_min)/(X_max-X_min + 1e-8)
```

```python
X
```

```python
X_scaled
```

```python
#the deep learning model works well with these valu
```

```python
 X_train, X_test, y_train, y_test = train_test_split( X_scaled, y, test_size=0.33, random_state=42)
```

```python
model = keras.Sequential([
    #first layer is input
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(8, activation='relu'),
    # ouutput layer
    layers.Dense(1, activation = 'softmax')
])
```

```python
# optimizer
model.compile(optimizer = 'sgd', loss = 'binary_crossentropy',metrics= ['accuracy'])
```

```python
history_full = model.fit(X_train,y_train,epochs = 100,batch_size = len(X_train),verbose = 1)

history = model.fit(
    X_train.values,y_train.values,
    validation_data = (X_test.values,y_test.values),
    epochs = 100,
    batch_size = 3,
    verbose = 1 # show what is being in data training
    #batch_size if we have 10k data points and it will take one row but if we put batch it willtaek rows in batches)
#save your training
)
history_minibatch = model.fit(X_train,y_train,epochs = 100,batch_size = 100,verbose = 4)
```
